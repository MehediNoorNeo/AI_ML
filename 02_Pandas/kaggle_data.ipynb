```python
import pandas as pd
```

```python
# AQI data set
df = pd.read_csv("globalAirQuality.csv")
```

```python
df.describe()
```

```python
# Selecting data
# selecting columns
df["country"]
df[["city", "aqi"]]
```

```python
# Selecting rows
# Select rows (by label)
df.loc[0]             # 1st row (by label)
df.loc[0:3]           # row0 to row3 - both inclusive in loc

# Select rows (by index)
df.iloc[0]            # 1st row 
df.iloc[4:7]          # row4 to row6 - ending row exclusive in iloc
```

```python
# Select rows & columns
df.loc[0, "country"]  # 0th row & specific column
df.iloc[0, 2]         # 1th row & 3rd column (by position)

df.iloc[0, 1:5]
df.loc[0, ["country", "aqi"]]

df.loc[0:5, ["country", "city", "latitude"]]   # end is inclusive
df.iloc[0:5, [1, 3, 6, 10]]                    # end is exclusive 
```

```python
# Select single scalar value
df.at[0, "city"]
df.iat[0, 2]
```

```python
# Filtering Data
df[df["aqi"]>100] #print rows where aqi > 100
df[df["aqi"]>100][["aqi", "city"]] #print rows where aqi > 100 and columns only aqi, city

df[(df["aqi"]>100) & (df["timestamp"] == "2025-11-04 18:25:17.554219")]
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] > 30)][["city", "aqi"]]


aqi_data.iloc[0]
# aqi_data.loc[0] # Error => not exist label 0 in aqi_data
aqi_data.loc[6] # show data of first row as like aqi_data.iloc.[0]

# Query - returns a copy, not a view
aqi_val = 100
df.query("aqi > 100")
df.query("aqi > 100 & temperature > 30")[["city", "aqi"]]
df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]]

```
