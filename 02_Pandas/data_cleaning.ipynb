```python
import pandas as pd
```

```python
# Data cleaning
df = pd.read_csv("raw_data.csv")
```

```python
df.isnull() # isna() is same function as isnull()
df.isnull().sum() # showed number of missing cell according to columns

df.dropna() # drop rows with missing values
df.dropna(axis = 1) # drop columns with missing values

df.fillna(0) # fills NaN with 0

# To save cleaning data
clean_data = df.copy()
clean_data["age"] = clean_data["age"].fillna(clean_data["age"].mean())
clean_data

df.ffill()             # forward fill with prev value
df.bfill()             # backward fill with prev value
```

```python
# Handle duplication
df.duplicated()  # True for duplicate rows
df["country"].duplicated() # True for duplicate rows in particular columns
df.duplicated(["country", "gender"])
df2 = df.copy()
df2.drop_duplicates(inplace = True)
df2
```

```python
# Data Types
df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64")
df2.dtypes

aqi_data = pd.read_csv("globalAirQuality.csv")
cpy_data = aqi_data.copy()
cpy_data["timestamp"] = pd.to_datetime(cpy_data["timestamp"])
cpy_data["timestamp"].dtypes
```

```python
# Handle string data
df["gender"].str.lower()
df["gender"].str.upper()
df["gender"].str.capitalize()

df["name"].str.split(" ")
# if data is mail ex: mehedi@gmail.com
# df["email"].str.split("@")

df["country"].str.contains("usa", case = False)
```

```python
# Data Transformation: Feature Engineering

df2 = df.copy()
# add new column
df2["tax"] = df2["income"].apply(lambda x: "20%" if x >= 60_000 else "10%")

# maping
gender_map = {"Male":"M", "Female" : "F", "Unknown": "U"}
df2["gender"] = df2["gender"].map(gender_map)

# assign function
df2 = df2.assign(new_income = df2["income"] * 1.1)

# replace(old, new) function
df2["country"] = df2["country"].replace("USA", "US")

# rename
df2.columns = ["Id", "Name", "Age", "Country", "Gender", "Income", "Tax", "New_Income"]
df2 = df2.rename(columns={"Income": "Salary"})
df2 = df2.rename(index={"first": 0})

# Sorting - values & index
df2["Age"] = df2["Age"].fillna(50)
df2.sort_values("Salary")
# df2.sort_values("Salary", ascending = False)
sorted_df2 = df2.sort_values(["Salary", "Age"])

sorted_df2.sort_index()

# reset index
sorted_df2.reset_index()
sorted_df2.reset_index(drop = True)

# Ranking
sorted_df2["Ranking"] = sorted_df2["Salary"].rank()
sorted_df2["Ranking"] = sorted_df2["Salary"].rank(ascending= False, method = "max") # dense, min

sorted_df2[["Id", "Name", "Age", "Country", "Gender", "Salary", "New_Income", "Ranking", "Tax"]] 
```

```python
# Grouping and Aggregation

df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].mean()
df.groupby("gender")["age"].min()

# Aggregation
df.groupby("country")["income"].agg(["mean", "min", "max"])
df.groupby("country")["income"].aggregate(["mean", "min", "max"])

df.groupby("country")["income"].aggregate(avg_income="mean", min_income="min", max_income="max")

df.groupby("country").agg({
    "income" : "mean",
    "age" : "min"
})
df.groupby("country").agg(
    avg_income=("income", "mean"),
    min_age=("age" , "min")
)
```

```python
# Melt and Pivot for Reshaping
# Melting
df4 = pd.DataFrame({
    "country": ["USA", "USA", "India", "India"],
    "year": [2020, 2021, 2020, 2021],
    "sales": [100, 120, 90, 110],
    "profit": [20, 25, 18, 22]
})

melted_df = df4.melt(
    id_vars=["country", "year"],     # columns to keep
    value_vars=["sales", "profit"],  # columns to unpivot or melting
    var_name="metric",               # new column name for variable
    value_name="value (in Lakhs)"    # new column name for value (default is value)
)
melted_df

# Pivoting
original_df = melted_df.pivot(
    index=["country", "year"],   # cols that become the NEW row index - can't for just country
    columns="metric",            # col whose unique values will become the cols
    values="value (in Lakhs)"    # col whose values will fill the new df
)
original_df
```
