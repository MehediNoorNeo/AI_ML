```python
import numpy as np
import time
```

```python
#execution performance
size = 1_000_000

#python list
py_list = list(range(size))
start = time.time()
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"python list time {end-start} second")

#numpy array 
np_array = np.array(py_list)
start = time.time()
sq_array = np_array**2
end = time.time()
print(f"numpy array time {end-start} second")
```

```python
# memory
import sys
print(f"Memory of python list {sys.getsizeof(py_list) + sys.getsizeof(py_list[0]) * len(py_list)} bytes")
print(f"Memory of numpy array {np_array.nbytes} bytes")
```
