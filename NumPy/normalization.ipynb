```python
import numpy as np
```

```python
# Standardization
arr = np.array([[1, 2], [3, 4]])
mean = np.mean(arr)
std_dev = np.std(arr)
print(mean)
print(std_dev)
standarized_arr = (arr-mean)/std_dev
print(standarized_arr)

print(np.mean(standarized_arr))
print(np.std(standarized_arr))

stzd = standarized_arr
print(np.sum(stzd**2)/stzd.size)
```
