```python
import pandas as pd
```

```python
# Data cleaning
df = pd.read_csv("raw_data.csv")
```

```python
df.isnull() # isna() is same function as isnull()
df.isnull().sum() # showed number of missing cell according to columns

df.dropna() # drop rows with missing values
df.dropna(axis = 1) # drop columns with missing values

# df.fillna(0) # fills NaN with 0

# To save cleaning data
clean_data = df.copy()
clean_data["age"] = clean_data["age"].fillna(clean_data["age"].mean())
clean_data

df.ffill()             # forward fill
df.bfill()             # backward fill
```

```python
# Handle duplication
df.duplicated()  # True for duplicate rows
df["country"].duplicated() # True for duplicate rows in particular columns
df.duplicated(["country", "gender"])
df2 = df.copy()
df2.drop_duplicates(inplace = True)
df2
```

```python
# Data Types
df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64")
df2.dtypes

aqi_data = pd.read_csv("globalAirQuality.csv")
cpy_data = aqi_data.copy()
cpy_data["timestamp"] = pd.to_datetime(cpy_data["timestamp"])
cpy_data["timestamp"].dtypes
```

```python
# Handle string data
df["gender"].str.lower()
df["gender"].str.upper()
df["gender"].str.capitalize()

df["name"].str.split(" ")
# if data is mail ex: mehedi@gmail.com
# df["email"].str.split("@")

df["country"].str.contains("usa", case = False)
```
