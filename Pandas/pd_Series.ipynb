```python
import pandas as pd
```

```python
# Serises
list = [4, "Akash", 4.5, "PK", 6]
s = pd.Series(list)
print(s)
print(type(s))

# index => position
print(s[3])
print(s[1])
```

```python
# label => name index
age = [34, 32, 23, 53, 22]
s = pd.Series(age, index=["Rohan", "Badsha", "Jamal", "Atis", "Asli"])
print(s)
print(s["Badsha"])
print(s["Atis"])
print(s[3])
print(s.index)
```

```python
# Homogenous => same data type(list: object)
# Vectorized operation
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([6, 7, 8, 9, 10])
print(s1+s2)

# Mutable values
s2[0] = 600
print(s2)

# Immutable size
chng_s1 = s1.drop(0)
print(s1)
print(chng_s1)
```
