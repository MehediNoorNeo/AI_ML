```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Temperature over cities
days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],   # New York
    [18, 19, 17, 20, 21],   # London
    [30, 32, 31, 33, 34],   # Delhi
    [25, 26, 24, 27, 28]    # Tokyo
]
fig, ax = plt.subplots(2, 2, figsize = (8, 6))

# city = 0;
# for i in range(2):
#     for j in range(2):
#         ax[i][j].plot(days, temperatures[city], label= cities[city], marker = "o")
#         ax[i][j].set_title(f"{cities[city]}'s temperatures")
#         ax[i][j].legend()
#         ax[i][j].grid()
#         city=city+1

axes = ax.flatten()

for i, ax in enumerate(axes): # idx, val at idx
    # print(f"{i} and {ax}")
    ax.plot(days, temperatures[i], marker='o')
    ax.set_title(cities[i])
    ax.grid(True)
    
fig.suptitle("Temperature in cities over the week")
fig.supylabel("Temperature (°C)")
fig.supxlabel("Day")
fig.tight_layout()
plt.show()
```
