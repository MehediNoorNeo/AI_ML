```python
import json
```

```python
# load the data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("store_data.json")
print(type(data), data)
```

```python
# clean and structure the data
def clean_data(data):
    text_to_num = {"one" : 1, "two" : 2, "three" : 3, "four" : 4, "five" : 5}
    clean_data = []
    unique_users = set()
    # rating cleaning
    for user in data:
        raw_rating = str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]
        user['rating'] = raw_rating
        #handle missing values
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None
        #deduplication
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        clean_data.append(user)
    return clean_data
```

```python
data = clean_data(data)
```

```python
print(data)
```

```python
# data dump
def dump_data(filename, data):
    with open(filename, "w") as f:
        json.dump(data, f, indent=4)
```

```python
dump_data("store_data.json", data)
```

```python
# get meaningful insights from data
def get_insights(data):
    #get avg rating
    total_rat = 0
    for user in data:
        total_rat += float(user["rating"])
    print(f"Average rating is {total_rat/len(data)}")
    # percentage of users with poor rating
    count = 0
    for user in data:
        if(float(user["rating"]) < 3):
            count += 1
    print(f"Percentage of poor rating is {(count/len(data))*100}")
        
```

```python
get_insights(data)
```

```python
# Build Recommendation feature
def build_rocommendation(data):
    recommendation = []
    for user in data:
        curr_recom = {}
        curr_recom["name"] = user["name"]
        if(float(user["rating"]) >= 4):
            curr_recom["brand"] = "Apple"
        else:
            curr_recom["brand"] = "Samsung"
        recommendation.append(curr_recom)
    return recommendation
    
```

```python
build_rocommendation(data)
```
